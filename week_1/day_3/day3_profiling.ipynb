```python
import pandas as pd

df = pd.read_csv("../incidents.csv")

df.head()
```

```python
df["opened_at"] = pd.to_datetime(df["opened_at"])

monthly_volume = df["opened_at"].dt.to_period("M").value_counts().sort_index()

print(monthly_volume)
```

```python
top_categories = df["category"].value_counts().head(10)

print(top_categories)
```

```python
resolution_by_priority = df.groupby("priority")["resolution_minutes"].agg(
    median="median",
    p90=lambda x: x.quantile(0.90)
)

print(resolution_by_priority)
```

```python
reassignment_stats = df["reassignment_count"].value_counts().sort_index()

print(reassignment_stats)
```

```python
sla_breach_count = df["sla_breached"].sum()
total_incidents = len(df)

sla_breach_rate = (sla_breach_count / total_incidents) * 100

print("SLA breaches:", sla_breach_count)
print("Total incidents:", total_incidents)
print("SLA breach rate:", sla_breach_rate, "%")
```

```python
sla_by_priority = df.groupby("priority")["sla_breached"].agg(
    breaches="sum",
    total="count"
)

sla_by_priority["breach_rate"] = (
    sla_by_priority["breaches"] / sla_by_priority["total"] * 100
)

print(sla_by_priority)
```

```python
group_counts = df["assignment_group"].value_counts()

print(group_counts)
```

```python
group_counts = df["assignment_group"].value_counts()

print(group_counts)
```
